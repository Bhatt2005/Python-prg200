# 2- Online Store Discount System

```python
# Karan Bhatt

# User Input
total = float(input("Enter total purchase amount (in NPR): "))
loyalty = input("Are you a loyalty member? (yes/no): ")

# Conditions

if total < 1000:
    discount = 0
elif total < 5000:
    discount = 5
elif total < 15000:
    discount = 10
else:
    discount = 20

# Applying discount

after_discount = total - (total * discount / 100)

# Applying loyalty discount

if loyalty == "yes":
    after_discount = after_discount - (after_discount * 5 / 100)

print("Final payable amount: NPR", after_discount)
```
