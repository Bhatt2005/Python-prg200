# 1- ATM Withdrawal Validator 

```python
# Karan Bhatt

# User Input

balance = int(input("Enter current account balance: "))
daily_withdrawn = int(input("Enter total amount withdrawn today: "))
amount = int(input("Enter amount to withdraw: "))

# Conditions

if amount % 500 != 0:
    print("Invalid amount.Amount must be a multiple of NPR 500.")

elif daily_withdrawn % 500 != 0:
    print("Invalid previous withdrawal record.")

elif amount > balance:
    print("Insufficient balance.")

elif daily_withdrawn + amount > 50000:
    print("Daily withdrawal limit reached.")

else:
    # Deducting both prev and current withdraws
    
    remaining_balance = balance - (daily_withdrawn + amount)

    print("Withdrawal successful.")
    print("Your current balance after withdrawal: NPR", remaining_balance)
```
