# 4- Shopping Discount

```python
#  Shopping Discount
# The assignment requires creating a separate module (discount.py) to practice
# splitting code across files and importing from a custom module.
# discount.py holds all the discount and tax logic, while this file only
# handles the products and calls the imported functions.

%%writefile discount.py

TAX_RATE = 0.13

def apply_discount(price, percent):
    return price - (price * percent / 100)

def apply_tax(price):
    return price + (price * TAX_RATE)

def final_price(price, discount_pct):
    discounted = apply_discount(price, discount_pct)
    return apply_tax(discounted)
```

```python


from discount import final_price, TAX_RATE

print(f"TAX_RATE imported from discount.py: {TAX_RATE}")
print()

products = [
    ("Laptop",      85000, 10),
    ("Headphones",   4500, 15),
    ("Phone Case",    800,  5),
    ("USB Cable",     600,  0),
]

for name, price, discount in products:
    total = final_price(price, discount)
    print(f"{name} | Original: NPR {price} | Final: NPR {total:.2f}")
```
