# 2-Bill Splitter 

```python
#  Bill Splitter
# Splits a restaurant bill equally among friends and picks one random person
#to pay an extra NPR 50 as a lucky tax.

import random

random.seed(42)

friends = ["Karan", "Bikram", "Rojan", "Sushant", "Dipendra"]
total_bill = 3750


def split_bill(friends, total):
    return total / len(friends)


# picks a random friend using random.choice

def pick_lucky(friends):
    return random.choice(friends)


def final_summary(friends, total):
    share = split_bill(friends, total)
    lucky = pick_lucky(friends)

    print(f"Bill per person: NPR {share:.2f}")
    print()

    for friend in friends:
        if friend == lucky:
            lucky_total = share + 50   
            print(f"{friend}: NPR {lucky_total:.2f} (lucky tax +NPR 50)")
        else:
            print(f"{friend}: NPR {share:.2f}")

    print(f"\n{lucky} was picked to pay the lucky tax!")


final_summary(friends, total_bill)
```
