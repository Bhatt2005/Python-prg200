# 1- Temperature Logger

```python
#  Temperature Logger

# Uses the math module to analyse daily temperature readings from a weather station.
# Calculates average, standard deviation, and prints a summary with min/max values.

import math

station_name = "Kathmandu Weather Station"

temperatures = [18.4, 22.1, 15.7, 29.3, 11.8, 25.6, 19.2]

# returns the mean 
def get_average(temps):
    return sum(temps) / len(temps)

# calculates standard deviation 
def get_deviation(temps):
    mean = get_average(temps)   
    variance = sum((t - mean) ** 2 for t in temps) / len(temps)
    return math.sqrt(variance)

# prints a full summary 

def get_summary(temps):
    print(f"Station: {station_name}")
    print(f"Min: {min(temps)} °C")
    print(f"Max: {max(temps)} °C")
    print(f"Average: {get_average(temps):.2f} °C")
    print(f"Standard Deviation: {get_deviation(temps):.2f}")


get_summary(temperatures)

```
