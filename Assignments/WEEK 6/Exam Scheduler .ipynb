# 3- Exam Scheduler 

```python
# Exam Scheduler
# Calculates and prints the exam date for each subject by adding days to a start date.

import datetime

college_name = "Bhaktapur Multiple Campus"

start_date = "2025-05-01"

exams = [
    ("Python Programming",  0),
    ("Data Structures",     3),
    ("Database Systems",    6),
    ("Computer Networks",  10),
    ("Mathematics",        14),
]


# converts a date string 'YYYY-MM-DD' into a datetime object
def parse_date(date_str):
    return datetime.datetime.strptime(date_str, "%Y-%m-%d")


# returns the exam date as a string by adding the given number of days to the start date

def get_exam_date(start_str, days):
    start = parse_date(start_str)
    exam_date = start + datetime.timedelta(days=days)
    return exam_date.strftime("%Y-%m-%d")


# prints the full exam schedule with college name as header

def print_schedule(start_str, exams):
    print(f"{college_name} — Exam Schedule")
    print("-" * 40)
    for subject, days in exams:
        date = get_exam_date(start_str, days)
        print(f"{subject}: {date}")


print_schedule(start_date, exams)
```
