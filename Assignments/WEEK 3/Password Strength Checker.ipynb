# 4- Password Strength Checker

```python
# Karan Bhatt

# User Input

password = input("Enter your password: ")

missing = []

#Conditions

if len(password) < 8:
    missing.append("At least 8 characters")

has_upper = False
for char in password:
    if char.isupper():
        has_upper = True
if not has_upper:
    missing.append("At least one uppercase letter")

has_lower = False
for char in password:
    if char.islower():
        has_lower = True
if not has_lower:
    missing.append("At least one lowercase letter")

has_digit = False
for char in password:
    if char.isdigit():
        has_digit = True
if not has_digit:
    missing.append("At least one digit")

has_special = False
for char in password:
    if char in "!@#$%^&*":
        has_special = True
if not has_special:
    missing.append("At least one special character (!@#$%^&*)")

if len(missing) == 0:
    print("Strong Password")
else:
    print("Weak Password. Missing:", missing)
```
