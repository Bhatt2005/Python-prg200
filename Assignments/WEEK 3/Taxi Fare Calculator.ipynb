# 5- Taxi Fare Calculator

```python
# Karan Bhatt

# User Input

distance = float(input("Enter distance (km): "))
hour = int(input("Enter travel hour (0-23): "))

# Base fare for first 2 km

fare = 150

# Next 8 km (km 3 to 10)
if distance > 2:
    if distance <= 10:
        fare = fare + (distance - 2) * 35
    else:
        fare = fare + 8 * 35

# Beyond 10 km

if distance > 10:
    fare = fare + (distance - 10) * 28

# Night surcharge (10 PM = 22, 5 AM = 5)

if hour >= 22 or hour < 5:
    fare = fare + fare * 10 / 100

print("Total Fare: NPR", fare)
```
