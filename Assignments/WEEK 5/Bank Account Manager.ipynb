# 1- Bank Account Manager

```python
# Bank Account Manager

class BankAccount:
    def __init__(self, name, account_number, balance=0):
        self.name = name
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited NPR {amount} to {self.name}. New balance: NPR {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Insufficient funds. {self.name} has only NPR {self.balance}, cannot withdraw NPR {amount}.")
        else:
            self.balance -= amount
            print(f"Withdrew NPR {amount} from {self.name}. New balance: NPR {self.balance}")

    def get_balance(self):
        print(f"{self.name}: NPR {self.balance}")


accounts = [
    ("Karan Bhatt",      "A001", 5000),
    ("Bikram Timilsina", "A002", 0),
    ("Rojan Tamang",     "A003", 12000),
]

a1 = BankAccount(*accounts[0])
a2 = BankAccount(*accounts[1])
a3 = BankAccount(*accounts[2])

a2.deposit(3000)
a3.withdraw(15000)   # This attempt fails because withrawl amount exceeds- 
a1.withdraw(2000)    # money present in the bank account

print("\n---- Final Balances ----")
a1.get_balance()
a2.get_balance()
a3.get_balance()
```
