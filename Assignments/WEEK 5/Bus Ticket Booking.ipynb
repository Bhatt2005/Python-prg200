# 4- Bus Ticket Booking

```python
# Bus Ticket Booking (Sajha Yatayat)

class Bus:
    def __init__(self, route, total_seats):
        self.route = route
        self.total_seats = total_seats
        self.booked = []

    # books a seat if available, otherwise notifies that it's already taken
    
    def book_seat(self, seat_number, passenger_name):
        for seat, _ in self.booked:
            if seat == seat_number:
                print(f"Seat {seat_number} already booked")
                return
        self.booked.append((seat_number, passenger_name))
        print(f"Seat {seat_number} booked for {passenger_name}")

    # returns the number of seats not yet booked
    
    def available_seats(self):
        return self.total_seats - len(self.booked)

    # prints all booked seats along with the passenger names
    def passenger_list(self):
        print(f"\nPassenger list — {self.route}")
        for seat, name in self.booked:
            print(f"  Seat {seat}: {name}")


bus = Bus("Kathmandu - Pokhara", 10)

bookings = [
    (3, "Karan Bhatt"),
    (7, "Sushant Ale Magar"),
    (3, "Bikram Timilsina"),   # duplicate
    (1, "Rojan Tamang"),
    (7, "Bibek Bhatt"),     # duplicate
]

for seat, name in bookings:
    bus.book_seat(seat, name)

print(f"\nAvailable seats: {bus.available_seats()}")
bus.passenger_list()
```
