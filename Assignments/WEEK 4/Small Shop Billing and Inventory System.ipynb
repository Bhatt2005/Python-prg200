# 1- Small Shop Billing and Inventory System

```python
# Small Shop Billing and Inventory System

inventory = {
    "rice":  {"price": 120, "stock": 20},
    "milk":  {"price": 90,  "stock": 10},
    "bread": {"price": 60,  "stock": 15},
    "eggs":  {"price": 15,  "stock": 30}
}

cart = {
    "rice": 12,
    "milk": 4,
    "eggs": 10
}


def process_order(inventory, cart):
    bill_items = {}
    grand_total = 0

    for item, quantity in cart.items():
        if item not in inventory:
            print(f"Sorry, '{item}' is not available in this shop.")
            continue

        if quantity > inventory[item]["stock"]:
            print(f"Sorry, not enough stock for {item}")
        else:
            item_total = inventory[item]["price"] * quantity
            bill_items[item] = {"quantity": quantity, "total": item_total}
            grand_total += item_total
            inventory[item]["stock"] -= quantity

    print("\n---- Bill ----")
    for item, details in bill_items.items():
        print(f"{item} x{details['quantity']} = NPR {details['total']}")
    print(f"Grand Total: NPR {grand_total}")
    print("--------------")

    updated = [f"{item}={info['stock']}" for item, info in inventory.items()]
    print("Updated stock:", ", ".join(updated))


process_order(inventory, cart)
```
