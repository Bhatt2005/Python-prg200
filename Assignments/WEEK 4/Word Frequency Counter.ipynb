# 4-Word Frequency Counter

```python
# Word Frequency Counter

text = """
Kathmandu is the capital city of Nepal. Kathmandu is full of temples and culture.
The city attracts many tourists every year. Tourists visit Kathmandu to explore
Pashupatinath and Boudhanath. Nepal is known for its culture and heritage.
Kathmandu remains the heart of Nepal.
"""


def word_frequency(text):
    text = text.lower()
    words = text.split()

    frequency = {}
    for word in words:
        word = word.strip(".,!?;:\"'")
        if word == "":
            continue
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    # sort by count (highest first) and grab top 3
    sorted_words = sorted(frequency.items(), key=lambda item: item[1], reverse=True)
    top_3 = sorted_words[:3]

    print("Top 3 words:")
    for word, count in top_3:
        print(f"{word} — {count} times")

    return top_3


word_frequency(text)
```
