# 3- Date Converter for Nepal Bank System 

```python
#  Date Converter for Nepal Bank System (BS <-> AD)
"""
Accurate month/day conversion requires a full BS calendar lookup table
which is outside the scope of this assignment
 """
bs_months = [
    "Baisakh", "Jestha", "Ashadh", "Shrawan", "Bhadra", "Ashwin",
    "Kartik", "Mangsir", "Poush", "Magh", "Falgun", "Chaitra"
]

customers = [
    {"name": "Karan Bhatt",       "date": "2000-03-15", "cal": "AD", "need": "BS", "style": "full"},
    {"name": "Bikram Timilsina",  "date": "2062-07-22", "cal": "BS", "need": "AD", "style": "iso"},
    {"name": "Sushant Ale Magar", "date": "2005-08-31", "cal": "AD", "need": "BS", "style": "nepali"},
    {"name": "Rojan Tamang",      "date": "2058-04-18", "cal": "BS", "need": "AD", "style": "full"},
]

def convert_date(date_str, from_cal, to_cal):
    if from_cal == to_cal:
        return date_str

    parts = date_str.split("-")
    year  = int(parts[0])
    month = parts[1]
    day   = parts[2]

    new_year = year + 56 if (from_cal == "AD" and to_cal == "BS") else year - 56
    return f"{new_year}-{month}-{day}"

def format_date(date_str, calendar, style):
    parts = date_str.split("-")
    year  = int(parts[0])
    month = int(parts[1])
    day   = int(parts[2])

    if style == "iso" or calendar == "AD":
        return f"{date_str} {calendar}"

    if 11 <= day <= 13:
        suffix = "th"
    elif day % 10 == 1:
        suffix = "st"
    elif day % 10 == 2:
        suffix = "nd"
    elif day % 10 == 3:
        suffix = "rd"
    else:
        suffix = "th"

    return f"{day}{suffix} {bs_months[month - 1]}, {year} {calendar}"

for customer in customers:
    converted = convert_date(customer["date"], customer["cal"], customer["need"])
    formatted = format_date(converted, customer["need"], customer["style"])
    print(f"{customer['name']} | Original: {customer['date']} {customer['cal']} | Converted: {formatted}")
```
